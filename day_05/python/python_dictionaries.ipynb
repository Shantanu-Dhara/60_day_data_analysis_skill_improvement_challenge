*Dictionary*

```python
# Basic syntax

a = {
    "Name": "Shantanu",
    "Age": 21,
    "Rumela": "Shantanu"
    }
print(a["Name"])
```

*Dictionary Functions*

```python
# Dictionary functions
a = {
    "Name": "Shantanu",
    "Age": 21,
    "Rumela": "Shantanu"
    }

b = a.get("Name")
print(b)

c = a.items()
print(c)

d = a.keys()
print(d)


e = a.values()
print(e)

f = a.copy()
print(f)
```

*Dictionary With Loop*

```python
# Dictionary with loop
a = {
    "Name": "Shantanu",
    "Age": 21,
    "Rumela": "Shantanu"
    }

for i in a:
    print(a[i])

for x, y in a .items():
    print(x,":", y)
```

*Nested Dictionary*

```python
a = {1: {
    "Name": "Shantanu",
    "Age": 21,
    "Rumela": "Shantanu"
    },
    2: 
    {
    "Name": "Shantanu",
    "Age": 21,
    "Rumela": "Shantanu"
    }}
print(a[1])
```
