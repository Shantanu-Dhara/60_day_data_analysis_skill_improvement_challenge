*Python Functions*

```python
#  Normal function
def R():
    print("Rumi")
R()

# Parameters and arguments
def hello(x, y):
    print(x + y)
hello(10, 20)
```

*Arbitary Arguments and Return statement*

```python
# Rbitary Arguments 
def Rbitary(*name):
    print("My name is: ",name[0])
Rbitary("Shantanu", "Rumi")

#  Return statement
def add(x, y):
    return x + y
print(add(10, 20))
```

*Factorial*

```python
# Write a program to calculate factorial of any number by using function
def fact(x):
    if x == 1:
        return 1
    else:
        return x*fact(x-1)
print(fact(5))
```

*Lamda Function*

```python
#  It's a anonomous sort term function 

a = lambda b: b*5
print(a(5))

x = lambda a, b, c: (a+b)*c
print(x(2, 3, 5))
```

*Local and Global Variable*

```python
# Local variable
x = 10
print("First variable is: ", x)

def hello():
    x = 11
    return x
print(hello())
print(x)

# Global variable
x = 10
print("First global variable is: ", x)

def hello():
    global x
    x = 11
    return x
print(hello())
print(x)
```
