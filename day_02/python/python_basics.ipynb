```python
# python basics of variable, data type, operators, conditional and for loop

#  Variables & Data Types
a = "Shantanu"
b = 10
c = 10.20
d = True

# Changing type of (c) float to integer
e = c
c = int(e)
print(type(a), type(b), type(c), type(d))
```

```python
#  Operators

f, g = 10, 5
print(f + g, f > g, f == g, not (f < g))
```

```python
#  Conditional Statements (if-else)

if 20 > 30:
    print("Yes")
elif 20 > 15:
    print("Exactly")
else:
    print("Not")

text = "Data Analysis"
print(text.lower(), text.upper(), text.split())

# basics of for loop
for i in range(5):
    print("Loop:", i)
```
