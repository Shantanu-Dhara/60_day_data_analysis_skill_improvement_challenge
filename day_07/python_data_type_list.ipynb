*List Funtions*

```python
# Length of a list

a = ["Shantanu Dhara", "Rumela Dhara", 10, 20]
print(len(a))

print("**********Break 1**********")
# Count an occurance of particular element

print(a.count("Shantanu Dhara"))
print("**********Break 2**********")

# Add to the list

a.append("Ritu Dhara")
print(a)
print("**********Break 3**********")


# Add to specific location

a.insert(0, "Bani")
print(a)
print("**********Break 4**********")

# Remove from a list
a.remove(10)
print(a)
print("**********Break 5**********")

# Remove from a certain location

a.pop(3)
print(a)
print("**********Break 6**********")

# Index find

a.index("Shantanu Dhara")
print(a)
print("**********Break 7**********")

# Extend a list

b = ["Bappa Dhara", "Sitesh Dhara"]
a.extend(b)
print(a)
print("**********Break 8**********")

# Reverse the list

a.reverse()
print(a)

# Sort of list

c = [5, 4, 9, 1, 72, 10, 55, 20, 98, 100]
c.sort()
print(c)
```

*Iteration Of List With Loop*

```python
# Simple iteration of list using for loop
a = ["Shantanu Dhara", "Rumela Dhara", 10, 20]

for i in a:
    print(i)

print("********Break 1********")

# Iteration using for loop with range and length

for i in range(len(a)):
    print(a[i]) 

print("********Break2 ********")

# Iteration using while loop 
i = 0
while i < len(a):
    print(a[i])
    i+= 1

print("********Break 3********")

# Short-hand for loop

[print(i) for i in a]
```

*List Comprehension*

```python
# It's used when one list data we want to copy in another
a = [30, 40, 50, 60]

b = [i for i in a if i > 40]
print(b)

```

*List Slicing*

```python
a = ["Shantanu Dhara", "Rumela Dhara", 10, 20]
print(a[::1])
print(a[-1])
print(a[0:3])
print(a[3:4])
print(a[1:2])
```
