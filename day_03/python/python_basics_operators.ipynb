*Arithmatic Operators*

```python
# Arithmatic Operators
a = 8
b = 4

# Addition
print(a + b)

# Substraction
print(a - b)

# Multiplication
print(a * b)

# Division
print(a / b)

# Floor Division
print(a // b)

# Exponentiation
print(a**b)
```

*Assignment Operators*

```python
# Assignment Operators
# 1) Plus Equal to 

x = 10
x+=20

print(x)


# 2) Minus Equal to 

y = 10
y -= 5

print(y)

# 3) Multiply Equal to


z = 10
z *=10

print(z)

# 4) Equal to

x = 10
y = 2

x = y
print(x)
```

*Comparison Operators*

```python
# Comparison Operators
# 1) Greater than 
x = 10
y = 20

print(x > y)

# 2) Less than 
x = 10
y = 20

print(x < y)

# 3) Greater than Equal to
x = 10
y = 20

print(x >= y)

# 4) Less than Equal to
x = 10
y = 20

print(x <= y)

# 5) Equal to  
x = 10
y = 20

print(x == y)

# 6) Not Equal to  
x = 10
y = 20

print(x != y)

```

*Logical Operators*

```python
# Logical Operators
# 1) and operator 
x = 10
y = 10

if x == y and x <= y:
    print("True")
else:
    print("False")

# 2) or operator 
x = 10
y = 10

if x == y or x < y:
    print("True")
else:
    print("False")

# 3) not operator 
x = 10
y = 10

if not(x > y or x < y):
    print("True")
else:
    print("False")

```

*Membership Operators*

```python
# Membership Operators
# 1) in operator

x = "Shantanu Dhara"

print("Dhara" in x )

# 2) not in operator

x = "Shantanu Dhara"

print("Q" not in x )
```
